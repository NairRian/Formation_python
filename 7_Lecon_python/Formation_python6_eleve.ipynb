### Création d'une variable

Une variable est un conteneur qui stocke une valeur. Pour créer une variable en Python, il suffit de lui donner un nom et de lui affecter une valeur.

## Listes

Les listes sont utilisées pour stocker une collection d'objets en Python.

Elles sont délimitées par des crochets **[]**.

*Tous types de données* peuvent être stockés dans une liste, y compris des listes imbriquées.

```python
student = ["Olivia", "Liam", "Ava", "Noah", "Sophia", "Ethan", "Isabella", "Mason", "Charlotte", "James"]
```

Les éléments d'une liste peuvent être accédés en utilisant leur *indice*, qui commence à 0 et va jusqu'à len(liste) - 1.

Les indices négatifs peuvent également être utilisés pour accéder aux éléments à partir de la fin de la liste.

La fonction len() peut être utilisée pour déterminer le **nombre d'éléments** dans une liste.

```python
# Nombre d'éléments dans une liste

print(len(student)) #10
```

```python
# trouver l'index d'un élément

student.index("Olivia")
```

```python
# Accéder aux éléments d'une liste

print(student[0])  # Olivia
```

```python
print(student[-1])  # James
```

```python
print(student[3:6]) # ['Noah', 'Sophia', 'Ethan']
```

```python
# vérifier la présence d'un élément dans la liste
"Isabella" in student
```

Les éléments d'une liste peuvent être **modifiés** en utilisant leur indice et en assignant une nouvelle valeur. (NON IMMUABLE)

```python
# Modifier un élément d'une liste

student[1] = "Abby"
print(student)
```

Les nouvelles valeurs peuvent être **ajoutées** à une liste en utilisant la méthode append() ou insert().

```python
# Ajouter un élément à une liste en dernière position

student.append("New Student")
print(student)
```

```python
# Ajouter un élément à une liste à une indexation précise
student.insert(2, "New Student 2")
print(student)
```

```python
# insérer plusieurs éléments à la fois en ajoutant une liste

student.extend(["Juliette", "Roméo"])

print(student)
```

Les éléments peuvent être **supprimés** d'une liste en utilisant la méthode remove() ou pop().

```python
# Supprimer un élément d'une liste

student.remove("Ava")
print(student)
```

```python
# supprimer un élément précis de la liste en indiquant son index
student.pop(2)
print(student)
```

```python
# supprimer le dernier élément de la liste
student.pop()
print(student)
```

La méthode sort() peut être utilisée pour **trier** les éléments d'une liste.

```python
# Trier les éléments d'une liste

student.sort()

print(student)
```

La méthode reverse() permet d'**inverser** l'ordre des éléments.

```python
student.reverse()
print(student)
```

La méthode **count()** permet d'avoir le nombre d'itération d'un élément dans une liste.

```python
student.count("Mason")
```

Les listes sont des collections ordonnées d'éléments. Elles peuvent contenir des éléments de **différents types**.

```python
mylist = [1, 2, 3, "Bonjour", True]
print(mylist)
```

Les éléments d'un tuple peuvent être accédés en utilisant leur *indice*, qui commence à 0 et va jusqu'à len(tuple) - 1.

Les tuples sont souvent utilisés pour stocker des paires de valeurs, telles que des coordonnées x, y.

### Créer un tuple à partir d'un autre

On **ne peut pas modifier un tuple** !
Toutefois, on peut en créer un à partir d'un autre.

```python
# ajouter des éléments d'un tuple à un autre

coordonnees = coordinates.__add__((4,))
print(coordinates)
print(coordonnees)
```

```python
# concaténer des tuples

tout = coordinates + coordonnees

tout
```

```python
# concaténer des fragments de tuples

partiel = coordinates[:5]+ coordonnees[3:]
partiel
```

```python
# répéter des tuples

multiple = coordinates*3

multiple
```

## Exercice 3 : If

Nous allons créer un système de vérifiaction de mot de passe
Vous savez, ces fameuses vérifications qui nous casse les pieds à chaque changement de mot de passe.

Le code doit vérifier que votre mot de passe contient plus de 8 caractères.

```python
password = input("Entrer votre mot de passe : ")
#Utilisation d'une nouvelle fonction 'len':
password_length = len(password)
#Vérifier si le mot de passe est inférieur à 8 caractères :
if password_length < 8:
    print("Mot de passe trop court !")
else:
    print("Mot de passe valide !")
```

Complexifions cette condition avec une condition intermédiaire : elif
Le code doit indiquer que si le nombre de caractères est entre 8 et 12, le mot de passe est faible.

```python
password = input("Entrer votre mot de passe : ")
#Utilisation d'une nouvelle fonction 'len':
password_length = len(password)
if password_length < 8:
    print("Mot de passe trop court !")
elif password_length >= 8 and password_length <= 12:   #Avec le raccourci 'ALT + ENTRÉE' on peut simplifier la condition.
    print("Mot de passe faible !")
else:
    print("Mot de passe valide !")
```

Découvrons l'intérêt d'utiliser des boucles

Rédiger un code qui permet d'afficher à l'écran tous les éléments d'une liste d'emails.

```python
liste = ["a@gmail.com", "b@gmail.com", "c@gmail.com", "d@gmail.com", "e@gmail.com"]
```

```python
print(len(liste))
```

```python
print(liste[0])
print(liste[1])
print(liste[2])
print(liste[3])
print(liste[4])
```

Pour éviter les redondances, il suffit de créer la boucle suivante :

*Rappel : boucle for quand on connait le nombre d'itérations, boucle while quand la fin dépend de la réalisation d'une condition*

```python
for i in range(len(liste)):
    print("Mail envoyé à " + liste[i])
```

Ce premier exemple parcourt tous les éléments de la liste selon leur index (de 0 à len()-1).

Python est capable de parcourir un(e) liste/tuple/dictionnaire par lui-même.

```python
for mail in liste :
    print("Mail envoyé à " + mail)
```

*ATTENTION*
Ici la variable mail est variable **locale** et non globale. C'est-à-dire qu'elle est définie uniquement dans la boucle et qu'on ne peut pas l'appeler en dehors. En effet, une fois que la boucle est finie, la variable est détruite.

### Module random

Random est un module intégré à Python qui permet de générer des nombres aléatoires. Il permet de faire appel à des fonctions spécifiques : random, randint, choice, shuffle

```python
import random
```

    Générer un nombre aléatoire:

La fonction principale du module random est random.random(). Elle renvoie un nombre flottant aléatoire compris entre 0 (inclus) et 1 (exclu).

```python
nombre_aleatoire = random.random()
print(nombre_aleatoire)  # Affiche un nombre décimal entre 0 et 1
```

    Générer un entier aléatoire:

Pour obtenir un entier aléatoire dans une plage spécifique, utilisez random.randint(a, b).

```python
nombre_entier = random.randint(1, 10)  # Génère un entier entre 1 et 10 (inclus)
print(nombre_entier)
```

    Choisir un élément aléatoire dans une liste:

La fonction random.choice(sequence) permet de sélectionner un élément aléatoire dans une liste, une chaîne de caractères ou un autre type de séquence.

```python
fruits = ["pomme", "banane", "orange"]
fruit_aleatoire = random.choice(fruits)
print(fruit_aleatoire)  # Affiche un fruit aléatoire de la liste
```

    Mélanger une liste aléatoirement:

La fonction random.shuffle(sequence) mélange les éléments d'une liste aléatoirement.

```python
cartes = ["As", "Roi", "Dame", "Valet", "10", "9", "8", "7", "6", "5", "4", "3", "2"]
random.shuffle(cartes)
print(cartes)  # Affiche la liste de cartes mélangée
```

### Module time

Le module time est un module intégré à Python qui permet de travailler avec le temps. Il offre des fonctions pour obtenir l'heure actuelle, mesurer des intervalles de temps, mettre en pause l'exécution d'un programme, etc.

Comment l'utiliser ?

    Importer le module:

Avant de pouvoir utiliser les fonctions de time, il faut l'importer dans votre code :

```python
import time
```

   **Obtenir l'heure actuelle:**

    time.time(): Renvoie le nombre de secondes écoulées depuis le 1er janvier 1970 (epoch) en tant que flottant.

```python
timestamp = time.time()
print(timestamp)
```

    time.localtime(): Renvoie une structure de données représentant l'heure locale actuelle.

```python
heure_locale = time.localtime()
print(heure_locale)
```

    time.strftime(format, t): Formate une date et une heure selon un format spécifié.

```python
heure_formatee = time.strftime("%H:%M:%S", time.localtime())
print(heure_formatee)  # Affiche l'heure au format HH:MM:SS
```

   **Mesurer des intervalles de temps:**

    time.sleep(seconds): Met en pause l'exécution du programme pendant un nombre spécifié de secondes.

```python
print("Début")
time.sleep(2)
print("Fin")  # S'affichera après 2 secondes
```

    Pour mesurer le temps d'exécution d'un bloc de code, vous pouvez utiliser time.time() avant et après le bloc et calculer la différence.

```python
debut = time.time()
# Code à mesurer
fin = time.time()
temps_execution = fin - debut
print(f"Temps d'exécution : {temps_execution} secondes")
```
