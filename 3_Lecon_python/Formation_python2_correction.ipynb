##    Introduction aux variables

### Création d'une variable

Une variable est un conteneur qui stocke une valeur. Pour créer une variable en Python, il suffit de lui donner un nom et de lui affecter une valeur.

Voici un exemple de code Python qui crée une variable :

```python
x = 5
```

### Modification d'une variable

Une variable peut être modifiée en lui affectant une nouvelle valeur.

Voici un exemple de code Python qui modifie une variable :

```python
x = 5
x = x + 1
print(x)  # 6
```

### Affichage d'une variable

Pour afficher la valeur d'une variable en Python, il suffit d'utiliser la fonction print() et de passer la variable en argument.

Voici un exemple de code Python qui affiche la valeur d'une variable :

```python
x = 5
print(x)  # 5
```

## Types de données

### Entiers

Les entiers sont des nombres entiers, sans virgule.

Voici un exemple de code Python qui utilise des entiers :

```python
x = 5
y = 10
z = x + y
print(z)  # 15
```

### Virgules flottantes

Les virgules flottantes sont des nombres à virgule.

Voici un exemple de code Python qui utilise des virgules flottantes :

```python
x = 5.5
y = 10.2
z = x + y
print(z)  # 15.7
```

### Chaînes de caractères

Les chaînes de caractères sont des séquences de caractères.

Voici un exemple de code Python qui utilise des chaînes de caractères :

```python
x = "Bonjour"
y = "monde"
z = x + " " + y
print(z)  # "Bonjour monde"
```

### Booléens

Les booléens sont des valeurs de vérité, qui peuvent être soit True, soit False.

Voici un exemple de code Python qui utilise des booléens :

```python
x = True
y = False
z = x and y
print(z)  # False
```

## Fonctions utiles

La fonction type() permet de connaître le type d'une variable.

Voici un exemple de code Python qui utilise la fonction type() :

### Fonction int()

La fonction int() permet de convertir une variable en entier.

Voici un exemple de code Python qui utilise la fonction int() :

Voici un exemple de code Python qui utilise la fonction float() :

### Fonction input()

La fonction input() permet de lire une valeur depuis l'entrée standard (clavier).

Voici un exemple de code Python qui utilise la fonction input() :

## Interactions avec les chaines de caractère/string

La chaîne Python est une collection ordonnée de caractères utilisée pour stocker et représenter des informations textuelles. Du point de vue fonctionnel, les chaînes peuvent être utilisées pour représenter à peu près tout ce qui peut être encodé en texte : symboles, mots, nombres, contenu de fichiers, etc.

Les chaînes sont créées en plaçant des guillemets (simples ou doubles) autour de la séquence de caractères donnée. Les chaînes sont des séquences immuables, ce qui signifie que l'ordre positionnel des caractères qu'elles contiennent, de gauche à droite, ne peut pas être modifié en place. Elles prennent en charge des opérations telles que :

    la concaténation (combinaison de chaînes)
    le découpage (extraction de sections)
    l'indexation (recherche par décalage)
    et la liste continue...

Python fournit également un ensemble de méthodes de chaîne qui mettent en œuvre des tâches spécifiques aux chaînes courantes, ainsi que des modules pour des tâches de traitement de texte plus avancées, telles que la correspondance de motifs. Créons quelques chaînes !

### Méthodes pour les string

Les méthodes sont simplement des fonctions associées à des objets particuliers. Tous les objets Python ont des méthodes intégrées, qui peuvent être appelées avec l'expression objet.méthode(arguments). Pour obtenir une liste de toutes les méthodes disponibles pour un objet spécifique, vous pouvez taper dir(objet) ou help(str) (pour chaîne). Les méthodes faciliteront les choses, je vous recommande de mémoriser les plus courantes.

Nous pouvons vérifier si un caractère se trouve dans une chaîne en utilisant **in**. Il existe également **not in**. Le résultat renvoie un booléen.

```python
prenom = "Patrick"

"c" in prenom
```

```python
"Pat" in prenom
```

```python
"T" not in prenom
```

Python offre de nombreuses fonctions pour manipuler les chaînes de caractères:

```python
mystring = "bonjour le monde !     "
```

Nous pouvons voir le nombre de caractères dans la chaîne en utilisant la fonction len() intégrée. Notez que l'espace compte comme un caractère.

```python
# combien de caractères dans la chaine ?
len(mystring)
```

Passer en majuscule le premier caractère de la liste

```python
print(mystring.capitalize())  # "Bonjour le monde!     "
```

Supprimer les espaces en début et fin de chaine

```python
print(mystring.strip())  # "Bonjour le monde!"
```

Passer tout en majuscule

```python
print(mystring.upper())  # "BONJOUR LE MONDE!"
```

Passer tout en minuscule

```python
print(mystring.lower())  # "BONJOUR LE MONDE!"
```

Compter le nombre d'itération d'un caractère dans une chaine

```python
print(mystring.count("o"))  # 3
```

Localiser un/des caractère(s) dans une chaine (fournit l'indice)

```python
print(mystring.find("monde"))  # 7
```

```python
print(mystring.find("!"))
```

Rappelez-vous que les chaînes sont immuables, nous ne pouvons donc pas les modifier en place. Par conséquent, l'opération suivante entraîne une erreur.

```python
nom_erreur = "Lusie"

# vous ne pouvez pas changer une seule valeur dans la chaine
nom_erreur[2] = "c"
```

On ne peut pas changer la variable, toutefois on peut en créer une autre qui serait corrigée à partir d'elle avec la méhode replace().

```python
nom_corrige = nom_erreur.replace('s','c')

nom_erreur, nom_corrige
```

### DECOUPAGE ET INDEXATION

Chaque chaîne peut être considérée comme un ordre de gauche à droite d'une séquence de caractères. Par conséquent, ces caractères peuvent être accédés via cet ordre. Pour les chaînes et autres objets Python, l'indexation commence à 0 et se termine à len()-1, et les indices que vous souhaitez accéder sont toujours entre crochets. Faisons du slicing !

```python
s = "spam"
```

```python
# premier caractère
s[0]
```

```python
# dernier caractère
s[-1]
```

Nous pouvons même découper des morceaux d'une chaîne en séparant les indices de début et de fin par un ":". L'indice de fin n'est pas inclus.

```python
salutations = "Bonjour à tous"
```

```python
# récupérer le premier mot (à partir de l'indexation 0 ou vide)

salutations[0:7], salutations[:7]
```

# Partie 3 : Opérations possibles

Vous vous souvenez de cet exemple ? Les chaines sont immuables et ne peuvent pas être remplacés.

```python
nom_erreur = "Lusie"

# vous ne pouvez pas changer une seule valeur dans la chaine
nom_erreur[2] = "c"
```

Pour l'instant, nous devrons effectuer cette correction via la concaténation. Ce sera notre premier exemple de polymorphisme ; l’idée que certains opérateurs agissent différemment selon le type d’objets sur lesquels ils agissent. Pour les nombres, l’opérateur « + » effectue une addition. Pour les chaînes, l’opération « + » concatène la chaîne. Notez que l’opérateur « + » pour les chaînes ne suit pas la propriété commutative car l’ordre a de l’importance. Donc, nous n’avons pas a + b = b + a si a et b sont des chaînes.

```python
trois_premiers = "abc"
trois_derniers = "def"

#concaténation
suite = trois_premiers + trois_derniers

suite
```

```python
# concaténation et découpage

prenom = "Albus"
deuxieme_prenom = "Perceval"
nom = "Dumbledore"

# récupérer les initiales
initiales = prenom[0]+deuxieme_prenom[0]+nom[0]

initiales
```

## Opérations arithmétiques et d'affectation

Les opérations arithmétiques en Python sont les suivantes :

    Addition : +
    Soustraction : -
    Multiplication : *
    Division : /
    Modulo : %
    Exponentiation : **
    Division entière : //
    Valeur absolue : abs()
    Racine carrée : sqrt()

Les opérations d'affectation en Python sont les suivantes :

    Affectation simple : =
    Affectation multiple : +=, -=, *=, /=, %=, **=, //=

Voici un exemple de code Python qui utilise des opérations arithmétiques et d'affectation :

```python
x = 5
x += 1
print(x)  # 6
```

```python
x = 5
x *= 2
print(x)  # 10
```

```python
x = 5
x /= 2
print(x)  # 2.5
```

```python
x = 5
x %= 2
print(x)  # 1
```

```python
x = 5
x **= 2
print(x)  # 25
```

```python
x = 5
x //= 2
print(x)  # 2
```

```python
#valeur absolue
abs(-4)
```

```python
#Importer depuis un package la méthode de la racine carrée
import math
math.sqrt(36)
```

## Opérateurs de comparaison

Les opérateurs de comparaison en Python sont les suivants :

    Égalité : ==
    Inégalité : !=
    Infériorité : <
    Supériorité : >
    Infériorité ou égalité : <=
    Supériorité ou égalité : >=

Voici un exemple de code Python qui utilise des opérateurs de comparaison :

```python
x = 5
y = 10
```

```python
print(x == y)  # False
```

```python
print(x != y)  # True
```

```python
print(x < y)  # True
```

```python
print(x > y)  # False
```

```python
print(x <= y)  # True
```

```python
print(x >= y)  # False
```
