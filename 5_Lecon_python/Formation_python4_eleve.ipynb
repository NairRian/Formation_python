##    Introduction aux variables

### Création d'une variable

Une variable est un conteneur qui stocke une valeur. Pour créer une variable en Python, il suffit de lui donner un nom et de lui affecter une valeur.

Voici un exemple de code Python qui crée une variable :

```python
x = 5
```

### Modification d'une variable

Une variable peut être modifiée en lui affectant une nouvelle valeur.

Voici un exemple de code Python qui modifie une variable :

```python
x = 5
x = x + 1
print(x)  # 6
```

### Affichage d'une variable

Pour afficher la valeur d'une variable en Python, il suffit d'utiliser la fonction print() et de passer la variable en argument.

Voici un exemple de code Python qui affiche la valeur d'une variable :

```python
x = 5
print(x)  # 5
```

## Types de données

### Entiers

Les entiers sont des nombres entiers, sans virgule.

Voici un exemple de code Python qui utilise des entiers :

```python
x = 5
y = 10
z = x + y
print(z)  # 15
```

### Virgules flottantes

Les virgules flottantes sont des nombres à virgule.

Voici un exemple de code Python qui utilise des virgules flottantes :

```python
x = 5.5
y = 10.2
z = x + y
print(z)  # 15.7
```

### Chaînes de caractères

Les chaînes de caractères sont des séquences de caractères.

Voici un exemple de code Python qui utilise des chaînes de caractères :

```python
x = "Bonjour"
y = "monde"
z = x + " " + y
print(z)  # "Bonjour monde"
```

### Booléens

Les booléens sont des valeurs de vérité, qui peuvent être soit True, soit False.

Voici un exemple de code Python qui utilise des booléens :

```python
x = True
y = False
z = x and y
print(z)  # False
```

## Fonctions utiles

### Fonction type()

La fonction type() permet de connaître le type d'une variable.

Voici un exemple de code Python qui utilise la fonction type() :

```python
x = 5
print(type(x))  # <class 'int'>
```

### Fonction str()

La fonction str() permet de convertir une variable en chaîne de caractères.

Voici un exemple de code Python qui utilise la fonction str() :

```python
x = 5
print(str(x))  # "5"
```

### Fonction int()

La fonction int() permet de convertir une variable en entier.

Voici un exemple de code Python qui utilise la fonction int() :

```python
x = "5"
print(int(x))  # 5
```

### Fonction float()

La fonction float() permet de convertir une variable en virgule flottante.

Voici un exemple de code Python qui utilise la fonction float() :

```python
x = "5.5"
print(float(x))  # 5.5
```

### Fonction input()

La fonction input() permet de lire une valeur depuis l'entrée standard (clavier).

## Interactions avec les chaines de caractère/string

La chaîne Python est une collection ordonnée de caractères utilisée pour stocker et représenter des informations textuelles. Du point de vue fonctionnel, les chaînes peuvent être utilisées pour représenter à peu près tout ce qui peut être encodé en texte : symboles, mots, nombres, contenu de fichiers, etc.

Les chaînes sont créées en plaçant des guillemets (simples ou doubles) autour de la séquence de caractères donnée. Les chaînes sont des séquences immuables, ce qui signifie que l'ordre positionnel des caractères qu'elles contiennent, de gauche à droite, ne peut pas être modifié en place. Elles prennent en charge des opérations telles que :

    la concaténation (combinaison de chaînes)
    le découpage (extraction de sections)
    l'indexation (recherche par décalage)
    et la liste continue...

Python fournit également un ensemble de méthodes de chaîne qui mettent en œuvre des tâches spécifiques aux chaînes courantes, ainsi que des modules pour des tâches de traitement de texte plus avancées, telles que la correspondance de motifs. Créons quelques chaînes !

Pour les nombres, l’opérateur « + » effectue une addition. Pour les chaînes, l’opération « + » concatène la chaîne. Notez que l’opérateur « + » pour les chaînes ne suit pas la propriété commutative car l’ordre a de l’importance. Donc, nous n’avons pas a + b = b + a si a et b sont des chaînes.

```python
trois_premiers = "abc"
trois_derniers = "def"

#concaténation
suite = trois_premiers + trois_derniers

suite
```

## Listes

Les listes sont utilisées pour stocker une collection d'objets en Python.

Elles sont délimitées par des crochets **[]**.

*Tous types de données* peuvent être stockés dans une liste, y compris des listes imbriquées.

```python
student = ["Olivia", "Liam", "Ava", "Noah", "Sophia", "Ethan", "Isabella", "Mason", "Charlotte", "James"]
```

Les éléments d'une liste peuvent être accédés en utilisant leur *indice*, qui commence à 0 et va jusqu'à len(liste) - 1.

Les indices négatifs peuvent également être utilisés pour accéder aux éléments à partir de la fin de la liste.

La fonction len() peut être utilisée pour déterminer le **nombre d'éléments** dans une liste.

```python
# Nombre d'éléments dans une liste

print(len(student)) #10
```

```python
# trouver l'index d'un élément

student.index("Olivia")
```

```python
# Accéder aux éléments d'une liste

print(student[0])  # Olivia
```

```python
print(student[-1])  # James
```

```python
print(student[3:6]) # ['Noah', 'Sophia', 'Ethan']
```

```python
# vérifier la présence d'un élément dans la liste
"Isabella" in student
```

Les éléments d'une liste peuvent être **modifiés** en utilisant leur indice et en assignant une nouvelle valeur. (NON IMMUABLE)

```python
# Modifier un élément d'une liste

student[1] = "Abby"
print(student)
```

Les nouvelles valeurs peuvent être **ajoutées** à une liste en utilisant la méthode append() ou insert().

```python
# Ajouter un élément à une liste en dernière position

student.append("New Student")
print(student)
```

```python
# Ajouter un élément à une liste à une indexation précise
student.insert(2, "New Student 2")
print(student)
```

```python
# insérer plusieurs éléments à la fois en ajoutant une liste

student.extend(["Juliette", "Roméo"])

print(student)
```

Les éléments peuvent être **supprimés** d'une liste en utilisant la méthode remove() ou pop().

```python
# Supprimer un élément d'une liste

student.remove("Ava")
print(student)
```

```python
# supprimer un élément précis de la liste en indiquant son index
student.pop(2)
print(student)
```

```python
# supprimer le dernier élément de la liste
student.pop()
print(student)
```

La méthode sort() peut être utilisée pour **trier** les éléments d'une liste.

```python
# Trier les éléments d'une liste

student.sort()

print(student)
```

La méthode reverse() permet d'**inverser** l'ordre des éléments.

```python
student.reverse()
print(student)
```

La méthode **count()** permet d'avoir le nombre d'itération d'un élément dans une liste.

```python
student.count("Mason")
```

Les listes sont des collections ordonnées d'éléments. Elles peuvent contenir des éléments de **différents types**.

```python
mylist = [1, 2, 3, "Bonjour", True]
print(mylist)
```

On peut créer une liste à partir d'une phrase avec la méthode **split()** autour d'un caractère précisé.

```python
# enregistrer chaque mot dans la liste

ligne = 'Je suis allée au restaurant.'

phrase = ligne.split(' ')
phrase
```

## Tuples

Les tuples sont similaires aux listes, mais ils sont **immuables**, ce qui signifie qu'ils *ne peuvent pas être modifiés une fois créés.

Les tuples sont délimités par des parenthèses **()**.

Les éléments d'un tuple peuvent être accédés en utilisant leur *indice*, qui commence à 0 et va jusqu'à len(tuple) - 1.

Les tuples sont souvent utilisés pour stocker des paires de valeurs, telles que des coordonnées x, y.

### Interactions possibles

```python
# Créer un tuple

coordinates = (1, 2, 3, 9, 6, 5, 2)

print(coordinates)
```

Il est possible de créer un tuple *à un élément*, toutefois il faut indiquer une virgule afin de s'assurer que cela ne soit pas confondu avec une variable

```python
tuple_simple = (4,)
tuple_simple
```

```python
# Accéder aux éléments d'un tuple

print(coordinates[0])  # 1
```

```python
print(coordinates[-1])  # 2
```

```python
# déterminer la longueur du tuple

len(coordinates)
```

```python
# compter le nombre d'occurence d'un élément
coordinates.count(9)
```

```python
# retourne l'index de la première occurence d'un élément
coordinates.index(2)
```

```python
# vérifier qu'un élément fasse partie du tuple

7 in coordinates
```

### Créer un tuple à partir d'un autre

On **ne peut pas modifier un tuple** !
Toutefois, on peut en créer un à partir d'un autre.

```python
# ajouter des éléments d'un tuple à un autre

coordonnees = coordinates.__add__((4,))
print(coordinates)
print(coordonnees)
```

```python
# concaténer des tuples

tout = coordinates + coordonnees

tout
```

```python
# concaténer des fragments de tuples

partiel = coordinates[:5]+ coordonnees[3:]
partiel
```

```python
# répéter des tuples

multiple = coordinates*3

multiple
```
