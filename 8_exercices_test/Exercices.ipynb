## **Exercice 1 : Organisation des tâches**

**Énoncé :**
Dans un bureau, les tâches à réaliser sont stockées dans une liste. Chaque jour, l’équipe réalise certaines de ces tâches.  
Écrivez un programme qui supprime progressivement les tâches réalisées de la liste et affiche l’état de la liste après chaque suppression.  
*imaginons que l'équipe effectue toujours la 1ère ou la dernière tâche de la liste*

**Indications :**

- Utiliser une boucle `while` pour supprimer les tâches.
- Afficher la `liste` après chaque suppression.

```python
# à vous
```

### **Correction :**

```python
# Liste des tâches
tasks = ["Réunion", "Rédaction de rapport", "Analyse de données", "Présentation client"]

while tasks:
    print("Tâches restantes :", tasks)
    done_task = tasks.pop(0)  # Suppression de la première tâche
    print(f"Tâche réalisée : {done_task}\n")
```

## **Exercice 2 : Suivi des ventes**

**Énoncé :**
Un commerçant enregistre ses ventes journalières dans une liste.   
Écrivez un programme qui calcule la somme totale des ventes et affiche chaque vente avec son numéro de jour.

**Indications :**

- Utiliser une boucle `for` pour parcourir les ventes.

La fonction `sum()` renvoie la somme des nombres que vous y renseignez.  
1 + 2 + 3  
6  
  
sum(1,2,3)  
6

La fonction `enumerate()` en Python est utilisée pour parcourir une séquence (comme une liste, un tuple ou une chaîne de caractères) tout *en gardant une trace de l’index de l’élément en cours d’itération*.   
  
for index, valeur in enumerate(sequence, start=0):
      
sequence : la collection à parcourir.  
start (optionnel) : l’index de départ (par défaut, 0).  
index : la position actuelle de l’élément dans la séquence.  
valeur : l’élément à cette position.  

```python
# à vous
```

### **Correction :**

```python
# Liste des ventes journalières
sales = [120, 250, 310, 90, 450]

total = sum(sales)
print(f"Total des ventes : {total}\n")

for sale in sales :
    day = sales.index(sale) +1
    print(f"Jour {day} : {sale} euros")
```

```python
# Liste des ventes journalières
sales = [120, 250, 310, 90, 450]

total = sum(sales)
print(f"Total des ventes : {total}\n")

for day, sale in enumerate(sales, start=1):
    print(f"Jour {day} : {sale} euros")
```

## **Exercice 5 : Vérification des mots de passe**

**Énoncé :**
Demandez à l’utilisateur de saisir son mot de passe et vérifiez s’il est correct en lui donnant 3 tentatives.  
*Le mot de passe sera déjà défini, le but est de vérifier si l'utilisateur s'en souvient.*

**Indications :**

- Utiliser une boucle `while` avec un compteur de tentatives.
- Comparer la saisie avec un mot de passe défini.

```python
# à vous
```

### **Correction :**

```python
password = "secure123"
attempts = 3

while attempts > 0:
    user_input = input("Entrez votre mot de passe : ")
    if user_input == password:
        print("Accès autorisé !")
        break
    else:
        attempts -= 1
        print(f"Mot de passe incorrect. Tentatives restantes : {attempts}")
```

## **Exercice 7 : Simulation d'une loterie**

**Énoncé :** 
Une entreprise organise une loterie pour ses employés. Chaque participant reçoit un numéro entre 1 et 50.   
Un programme doit tirer au hasard un numéro gagnant et vérifier s'il fait partie de la liste des participants.

**Indications :**

- Utiliser le module random pour tirer un numéro au hasard.
- Vérifier si ce numéro est dans la liste des participants.
- Afficher un message indiquant si quelqu'un a gagné ou non.
- *Indiquer le nom de la personne qui a gagné, donc faire un dictionaire au lieu d'une liste.*

```python
# à vous
```

### **Correction :**

```python
import random

participants = [3, 7, 15, 22, 29, 35, 41, 48]
gagnant = random.randint(1, 50)

print(f"Le numéro gagnant est : {gagnant}")
if gagnant in participants:
    print("Félicitations ! Un employé a gagné la loterie.")
else:
    print("Dommage, aucun gagnant cette fois-ci.")
```

```python
import random

participants = {"Patrick" : 3, "Virginie" : 7, "Lucien" : 15, "Régis" : 22, "Martine" :29, "Pauline" : 35}
gagnant = random.randint(1, 50)

print(f"Le numéro gagnant est : {gagnant}")

gagnant_nom = None

for nom, numero in participants.items():
    if numero == gagnant:
        gagnant_nom = nom
        break  # On arrête la boucle dès qu'on trouve le gagnant

if gagnant_nom:
    print(f"Félicitations ! {gagnant_nom} a gagné la loterie.")
else:
    print("Dommage, aucun gagnant cette fois-ci.")
```

## **Exercice 8 : Analyse de texte**

**Énoncé :**
Un responsable veut analyser un texte pour compter le nombre de lettres utilisées. Ce comptage permet d’étudier la fréquence des lettres pour améliorer la lisibilité des documents.

**Indications :**

- Supprimer ou ignorer les espaces du texte.
- Utiliser un dictionnaire pour stocker la fréquence des lettres.
- *Attention aux majuscules et minuscules*

Vous pouvez parcourir un dictionnaire en utilisant une boucle `for` et la méthode `.items()` pour récupérer à la fois les clés et les valeurs.  
for cle, valeur in dictionnaire.items() :

```python
# à vous
```

### **Correction :**

```python
texte = "Analyse de données pour un rapport"
compte_lettres = {}

for lettre in texte.replace(" ", "").lower(): #suppression des espaces et tout mis en minuscule
    compte_lettres[lettre] = compte_lettres.get(lettre, 0) + 1

for lettre, repet in compte_lettres.items() :
    print(f"La lettre {lettre} est apparue {repet} fois.")
```

```python
texte = "Analyse de données pour un rapport"
compte_lettres = {}

for lettre in texte:
    if lettre != " " : # ignorer les espaces et conservation des majuscules
        compte_lettres[lettre] = compte_lettres.get(lettre, 0) + 1

for lettre, repet in compte_lettres.items() :
    print(f"La lettre {lettre} est apparue {repet} fois.")
```

## **Exercice 9 : Gestion des stocks et livraisons**

**Énoncé :**
Un magasin doit livrer des produits à ses clients.   
Demandez à l’utilisateur quel produit et quelle quantité il souhaite livrer, puis mettez à jour le stock.

**Indications :**

- Vérifier si le produit existe et si le stock est suffisant.
- Mettre à jour le stock après la livraison.

```python
# à vous
```

### **Correction :**

```python
stock = {"Stylo": 20, "Cahier": 10, "Ordinateur": 5}
produit = input("Produit à livrer : ")
nombre = int(input("Quantité à livrer : "))

if produit in stock and stock[produit] >= nombre:
    stock[produit] -= nombre
    print(f"Livraison effectuée. Stock restant de {produit} : {stock[produit]}")
else:
    print("Stock insuffisant ou produit inexistant.")
```

```python
stock = {"Stylo": 20, "Cahier": 10, "Ordinateur": 5}
produit = input("Produit à livrer : ")
nombre = int(input("Quantité à livrer : "))

if produit in stock and stock[produit] >= nombre:
    stock[produit] -= nombre
    print(f"Livraison effectuée. Stock restant de {produit} : {stock[produit]}")
elif produit not in stock :
    print("Produit inexistant.")
elif stock[produit] < nombre :
    print("Stock insuffisant.")
```

## **Exercice 10 : Suivi des heures de travail**

**Énoncé :**
Un employeur souhaite vérifier si les employés respectent leur contrat d'heures de travail hebdomadaires.  
Ecrire un programme qui compare les heures effectuées par semaine des employés au quota horaire de leur contrat.

**Indications :**

- Stocker les heures travaillées par employé sous forme de liste.
- Comparer avec les heures du contrat.

```python
# à vous
```

### **Correction :**

```python
heures_travail = {
    "Alice": [35, 35], # [heures à faire, heures effectuées]
    "Bob": [40, 38],
    "Charlie": [30, 30],
    "Diane": [37, 40]
}

for employe, heures in heures_travail.items():
    heures_contrat, heures_effectuees = heures
    if heures_effectuees == heures_contrat:
        print(f"{employe} a respecté ses {heures_contrat} heures de travail.")
    elif heures_effectuees > heures_contrat:
        print(f"{employe} a dépassé ses heures de {heures_effectuees - heures_contrat} heures.")
    else:
        print(f"{employe} n'a pas complété ses heures, il manque {heures_contrat - heures_effectuees} heures.")
```
