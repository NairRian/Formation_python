##    INTRODUCTION AUX VARIABLES

### Création d'une variable

Une variable est un conteneur qui stocke une valeur. Pour créer une variable en Python, il suffit de lui donner un nom et de lui affecter une valeur.

Voici un exemple de code Python qui crée une variable :

```python
x = 5
```

### Modification d'une variable

Une variable peut être modifiée en lui affectant une nouvelle valeur.

Voici un exemple de code Python qui modifie une variable :

```python
x = 5
x = x + 1
print(x)  # 6
```

### Affichage d'une variable

Pour afficher la valeur d'une variable en Python, il suffit d'utiliser la fonction print() et de passer la variable en argument.

Voici un exemple de code Python qui affiche la valeur d'une variable :

```python
x = 5
print(x)  # 5
```

## Types de données

### Entiers

Les entiers sont des nombres entiers, sans virgule.

Voici un exemple de code Python qui utilise des entiers :

```python
x = 5
y = 10
z = x + y
print(z)  # 15
```

### Virgules flottantes

Les virgules flottantes sont des nombres à virgule.

Voici un exemple de code Python qui utilise des virgules flottantes :

```python
x = 5.5
y = 10.2
z = x + y
print(z)  # 15.7
```

### Chaînes de caractères

Les chaînes de caractères sont des séquences de caractères.

Voici un exemple de code Python qui utilise des chaînes de caractères :

```python
x = "Bonjour"
y = "monde"
z = x + " " + y
print(z)  # "Bonjour monde"
```

### Booléens

Les booléens sont des valeurs de vérité, qui peuvent être soit True, soit False.

Voici un exemple de code Python qui utilise des booléens :

```python
x = True
y = False
z = x and y
print(z)  # False
```

## Fonctions utiles

### Fonction type()

La fonction type() permet de connaître le type d'une variable.

Voici un exemple de code Python qui utilise la fonction type() :

```python
x = 5
print(type(x))  # <class 'int'>
```

### Fonction str()

La fonction str() permet de convertir une variable en chaîne de caractères.

Voici un exemple de code Python qui utilise la fonction str() :

```python
x = 5
print(str(x))  # "5"
```

### Fonction int()

La fonction int() permet de convertir une variable en entier.

Voici un exemple de code Python qui utilise la fonction int() :

```python
x = "5"
print(int(x))  # 5
```

### Fonction float()

La fonction float() permet de convertir une variable en virgule flottante.

Voici un exemple de code Python qui utilise la fonction float() :

```python
x = "5.5"
print(float(x))  # 5.5
```

### Fonction input()

La fonction input() permet de lire une valeur depuis l'entrée standard (clavier).

Voici un exemple de code Python qui utilise la fonction input() :

```python
x = input("Entrez une valeur : ")
print("Vous avez entré : " + x)
```
